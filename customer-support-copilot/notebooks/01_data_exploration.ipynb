## Day 1 Conclusion

The dataset contains multilingual customer support tickets in English and German.

Two useful prediction targets are available:

1. 'type': represents the nature of the ticket, such as Incident, Request, Problem, or Change.
2. 'queue': represents the team or department that should handle the ticket, such as Technical Support, Billing and Payments, or Sales and Pre-Sales.

For the baseline system, I will use 'subject + body' as input text.

Two baseline models will be built:

- Ticket Type Classifier: 'subject + body → type'
- Ticket Queue Classifier: 'subject + body → queue'

The 'answer' column will not be used as input because it represents the support response and can cause data leakage.

The 'priority' column will be used later for urgency detection.
