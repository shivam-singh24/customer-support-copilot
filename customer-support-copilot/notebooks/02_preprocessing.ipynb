```python
import pandas as pd
import re
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("../data/raw/tickets.csv")
```

```python
df = df[["subject", "body", "type", "queue", "priority", "language"]].copy()
```

```python
df["subject"] = df["subject"].fillna("")
df["body"] = df["body"].fillna("")
```

```python
df["text"] = df["subject"] + " " + df["body"]
```

```python
print(type(df['text'][0]))
```

```python
def clean_text(text):
    text = str(text).lower()
    text = re.sub(r"http\S+|www\S+", "", text)
    text = re.sub(r"\s+", " ", text)
    text = text.strip()
    return text
```

```python
df["clean_text"] = df["text"].apply(clean_text)
```

```python
df = df[df["clean_text"].str.len() > 0]
```

```python
train_df, test_df = train_test_split(
    df[["clean_text", "type", "queue", "priority", "language"]],
    test_size=0.2,
    random_state=42,
    stratify=df["type"]
)
```

```python
train_df.to_csv("../data/processed/train.csv", index=False)
test_df.to_csv("../data/processed/test.csv", index=False)
```

```python
type_names = sorted(df["type"].unique())

type_mapping = pd.DataFrame({
    "class_id": range(len(type_names)),
    "type": type_names
})

type_mapping.to_csv("../data/processed/type_mapping.csv", index=False)

type_mapping
```

```python
queue_names = sorted(df["queue"].unique())

queue_mapping = pd.DataFrame({
    "class_id": range(len(queue_names)),
    "queue": queue_names
})

queue_mapping.to_csv("../data/processed/queue_mapping.csv", index=False)

queue_mapping
```
